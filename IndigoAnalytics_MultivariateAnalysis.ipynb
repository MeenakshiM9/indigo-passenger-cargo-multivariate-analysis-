## STEP 1 Import libraries and load the workbook

```python
# STEP 1: IMPORT LIBRARIES AND LOAD THE EXCEL WORKBOOK


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

# ------------------------------------------------------------
# File location
# ------------------------------------------------------------

file_path = Path("data") / "Indigoairlines_data_updated.xlsx"

# ------------------------------------------------------------
# Check whether the file exists
# ------------------------------------------------------------

print("File exists:", file_path.exists())

# ------------------------------------------------------------
# Load the Excel workbook
# ------------------------------------------------------------

excel_file = pd.ExcelFile(file_path)

# ------------------------------------------------------------
# Display workbook information
# ------------------------------------------------------------

print("\nAvailable yearly sheets:")
print(excel_file.sheet_names)

print("\nNumber of yearly sheets:", len(excel_file.sheet_names))

print("\nYears covered:")
print(
    excel_file.sheet_names[0],
    "to",
    excel_file.sheet_names[-1]
)
```

## STEP 2 Inspect the sheet structure

```python
# STEP 2: INSPECT REPRESENTATIVE YEARLY SHEETS


years_to_check = ["2009", "2019", "2020", "2025"]

for year in years_to_check:

    print("\n" + "=" * 70)
    print(f"SHEET: {year}")
    print("=" * 70)

    df_temp = pd.read_excel(
        file_path,
        sheet_name=year,
        header=None
    )

    print(f"Shape: {df_temp.shape}")

    display(df_temp.head(10))
```

## STEP 3 Combine all yearly sheets

```python
# STEP 3: COMBINE ALL YEARLY SHEETS


# Standardized column names based on the DGCA data structure

column_names = [
    "Month",
    "Aircraft_Departures",
    "Aircraft_Hours",
    "Aircraft_Km",
    "Passengers_Carried",
    "Passenger_Km",
    "Available_Seat_Km",
    "Passenger_Load_Factor",
    "Freight_Tonnes",
    "Mail_Tonnes",
    "Total_Cargo_Tonnes",
    "Passenger_Tonne_Km",
    "Freight_Tonne_Km",
    "Mail_Tonne_Km",
    "Total_Tonne_Km",
    "Available_Tonne_Km",
    "Weight_Load_Factor"
]

# Month names that represent valid observations
valid_months = [
    "JAN", "FEB", "MAR", "APR", "MAY", "JUNE",
    "JULY", "AUG", "SEP", "OCT", "NOV", "DEC"
]

month_numbers = {
    "JAN": 1,
    "FEB": 2,
    "MAR": 3,
    "APR": 4,
    "MAY": 5,
    "JUNE": 6,
    "JULY": 7,
    "AUG": 8,
    "SEP": 9,
    "OCT": 10,
    "NOV": 11,
    "DEC": 12
}

# Store cleaned yearly data
all_years = []

# ------------------------------------------------------------
# Read and clean every yearly sheet
# ------------------------------------------------------------

for year in excel_file.sheet_names:

    df_year = pd.read_excel(
        file_path,
        sheet_name=year,
        header=None
    )

    # Keep rows where the first column contains a valid month.
    # This automatically handles the different header layouts.
    df_year["Month_Check"] = (
        df_year.iloc[:, 0]
        .astype(str)
        .str.strip()
        .str.upper()
    )

    df_year = df_year[
        df_year["Month_Check"].isin(valid_months)
    ].copy()

    # Remove temporary column
    df_year = df_year.drop(columns="Month_Check")

    # Keep the original 17 variables
    df_year = df_year.iloc[:, :17].copy()

    # Assign standardized names
    df_year.columns = column_names

    # Standardize month names
    df_year["Month"] = (
        df_year["Month"]
        .astype(str)
        .str.strip()
        .str.upper()
    )

    # Add year
    df_year["Year"] = int(year)

    # Add month number
    df_year["Month_Number"] = df_year["Month"].map(month_numbers)

    # Create date
    df_year["Date"] = pd.to_datetime(
        dict(
            year=df_year["Year"],
            month=df_year["Month_Number"],
            day=1
        ),
        errors="coerce"
    )

    # Convert operational variables to numeric
    numeric_columns = column_names[1:]

    for column in numeric_columns:
        df_year[column] = pd.to_numeric(
            df_year[column],
            errors="coerce"
        )

    all_years.append(df_year)


# ------------------------------------------------------------
# Combine all years
# ------------------------------------------------------------

airline_data = pd.concat(
    all_years,
    ignore_index=True
)

# ------------------------------------------------------------
# Arrange columns
# ------------------------------------------------------------

airline_data = airline_data[
    [
        "Date",
        "Year",
        "Month",
        "Month_Number"
    ] + numeric_columns
]

# Sort chronologically
airline_data = (
    airline_data
    .sort_values("Date")
    .reset_index(drop=True)
)

# ============================================================
# BASIC VERIFICATION
# ============================================================

print("Combined dataset shape:", airline_data.shape)

print("\nDate range:")
print(
    airline_data["Date"].min(),
    "to",
    airline_data["Date"].max()
)

print("\nObservations by year:")
print(
    airline_data.groupby("Year").size()
)

print("\nFirst 5 observations:")
display(airline_data.head())

print("\nLast 5 observations:")
display(airline_data.tail())
```

## STEP 4 Data-quality and missing-value assessment

```python
# STEP 4: DATA-QUALITY AND MISSING-VALUE ASSESSMENT


# ------------------------------------------------------------
# 4.1 Dataset dimensions and data types
# ------------------------------------------------------------

print("=" * 70)
print("DATASET DIMENSIONS")
print("=" * 70)

print("Rows    :", airline_data.shape[0])
print("Columns :", airline_data.shape[1])

print("\n" + "=" * 70)
print("DATA TYPES")
print("=" * 70)

print(airline_data.dtypes)


# ------------------------------------------------------------
# 4.2 Missing values
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("MISSING VALUES")
print("=" * 70)

missing_summary = pd.DataFrame({
    "Missing_Count": airline_data.isna().sum(),
    "Missing_Percentage": (
        airline_data.isna().mean() * 100
    ).round(2)
})

# Show only variables containing missing values
display(
    missing_summary[
        missing_summary["Missing_Count"] > 0
    ]
)


# ------------------------------------------------------------
# 4.3 Duplicate observations
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DUPLICATE CHECK")
print("=" * 70)

duplicate_count = airline_data.duplicated().sum()

print(
    "Number of completely duplicated rows:",
    duplicate_count
)


# ------------------------------------------------------------
# 4.4 Duplicate dates
# ------------------------------------------------------------

duplicate_dates = airline_data["Date"].duplicated().sum()

print(
    "Number of duplicated dates:",
    duplicate_dates
)


# ------------------------------------------------------------
# 4.5 Expected vs observed months
# ------------------------------------------------------------

expected_months = pd.date_range(
    start=airline_data["Date"].min(),
    end=airline_data["Date"].max(),
    freq="MS"
)

observed_months = pd.DatetimeIndex(
    airline_data["Date"].dropna().unique()
)

missing_dates = expected_months.difference(
    observed_months
)

print("\n" + "=" * 70)
print("EXPECTED VS OBSERVED MONTHS")
print("=" * 70)

print(
    "Expected monthly observations:",
    len(expected_months)
)

print(
    "Observed monthly observations:",
    len(observed_months)
)

print("\nMissing calendar months:")

if len(missing_dates) == 0:
    print("None")
else:
    for date in missing_dates:
        print(date.strftime("%Y-%m"))


# ------------------------------------------------------------
# 4.6 Descriptive statistics
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DESCRIPTIVE STATISTICS")
print("=" * 70)

display(
    airline_data[numeric_columns]
    .describe()
    .T
    .round(2)
)
```

## STEP 5 Range and logical consistency checks

```python
# STEP 5: RANGE AND LOGICAL CONSISTENCY CHECKS


print("=" * 70)
print("RANGE CHECKS")
print("=" * 70)

# ------------------------------------------------------------
# 5.1 Check for negative values
# ------------------------------------------------------------

negative_counts = (
    airline_data[numeric_columns] < 0
).sum()

print("\nNegative-value check:")

display(
    negative_counts[
        negative_counts > 0
    ].to_frame("Negative_Count")
)


# ------------------------------------------------------------
# 5.2 Check load-factor ranges
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("LOAD-FACTOR RANGE CHECK")
print("=" * 70)

plf_invalid = airline_data[
    (airline_data["Passenger_Load_Factor"] < 0) |
    (airline_data["Passenger_Load_Factor"] > 100)
]

wlf_invalid = airline_data[
    (airline_data["Weight_Load_Factor"] < 0) |
    (airline_data["Weight_Load_Factor"] > 100)
]

print(
    "Invalid Passenger Load Factor values:",
    len(plf_invalid)
)

print(
    "Invalid Weight Load Factor values:",
    len(wlf_invalid)
)


# ------------------------------------------------------------
# 5.3 Cargo consistency check
# Total Cargo = Freight + Mail
# ------------------------------------------------------------

cargo_check = airline_data[
    ["Freight_Tonnes", "Mail_Tonnes", "Total_Cargo_Tonnes"]
].copy()

cargo_check["Calculated_Total_Cargo"] = (
    cargo_check["Freight_Tonnes"] +
    cargo_check["Mail_Tonnes"]
)

cargo_check["Difference"] = (
    cargo_check["Total_Cargo_Tonnes"] -
    cargo_check["Calculated_Total_Cargo"]
)

cargo_inconsistent = cargo_check[
    cargo_check["Difference"].abs() > 1e-6
]

print("\n" + "=" * 70)
print("CARGO CONSISTENCY CHECK")
print("=" * 70)

print(
    "Rows with inconsistent Total_Cargo_Tonnes:",
    len(cargo_inconsistent)
)


# ------------------------------------------------------------
# 5.4 Tonne-km consistency check
# Total Tonne-Km =
# Passenger Tonne-Km + Freight Tonne-Km + Mail Tonne-Km
# ------------------------------------------------------------

tonne_check = airline_data[
    [
        "Passenger_Tonne_Km",
        "Freight_Tonne_Km",
        "Mail_Tonne_Km",
        "Total_Tonne_Km"
    ]
].copy()

tonne_check["Calculated_Total_Tonne_Km"] = (
    tonne_check["Passenger_Tonne_Km"] +
    tonne_check["Freight_Tonne_Km"] +
    tonne_check["Mail_Tonne_Km"]
)

tonne_check["Difference"] = (
    tonne_check["Total_Tonne_Km"] -
    tonne_check["Calculated_Total_Tonne_Km"]
)

tonne_inconsistent = tonne_check[
    tonne_check["Difference"].abs() > 1e-6
]

print("\n" + "=" * 70)
print("TONNE-KM CONSISTENCY CHECK")
print("=" * 70)

print(
    "Rows with inconsistent Total_Tonne_Km:",
    len(tonne_inconsistent)
)


# ------------------------------------------------------------
# 5.5 Available capacity checks
# ------------------------------------------------------------

seat_capacity_invalid = airline_data[
    airline_data["Available_Seat_Km"] < 0
]

tonne_capacity_invalid = airline_data[
    airline_data["Available_Tonne_Km"] < 0
]

print("\n" + "=" * 70)
print("CAPACITY CHECK")
print("=" * 70)

print(
    "Negative Available_Seat_Km values:",
    len(seat_capacity_invalid)
)

print(
    "Negative Available_Tonne_Km values:",
    len(tonne_capacity_invalid)
)


# ------------------------------------------------------------
# 5.6 Overall quality summary
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("OVERALL QUALITY SUMMARY")
print("=" * 70)

print("Duplicate rows:", airline_data.duplicated().sum())
print("Duplicate dates:", airline_data["Date"].duplicated().sum())
print("Negative-value variables:",
      (negative_counts > 0).sum())
print("Invalid Passenger Load Factor rows:",
      len(plf_invalid))
print("Invalid Weight Load Factor rows:",
      len(wlf_invalid))
print("Inconsistent cargo rows:",
      len(cargo_inconsistent))
print("Inconsistent tonne-km rows:",
      len(tonne_inconsistent))
print("Missing calendar months:",
      len(missing_dates))
```

## STEP 7 Define the multivariate analysis variables

```python
# STEP 7: DEFINE CORE MULTIVARIATE ANALYSIS VARIABLES


pca_variables = [
    "Aircraft_Departures",
    "Passengers_Carried",
    "Total_Cargo_Tonnes",
    "Available_Seat_Km",
    "Passenger_Load_Factor",
    "Weight_Load_Factor"
]

print("=" * 70)
print("SELECTED MULTIVARIATE ANALYSIS VARIABLES")
print("=" * 70)

for i, variable in enumerate(pca_variables, start=1):
    print(f"{i}. {variable}")

print("\nNumber of variables:", len(pca_variables))


# ------------------------------------------------------------
# Create analysis-ready complete-case dataset
# ------------------------------------------------------------

analysis_data = airline_data[
    ["Date", "Year", "Month"] + pca_variables
].dropna(
    subset=pca_variables
).copy()

analysis_data = analysis_data.sort_values(
    "Date"
).reset_index(drop=True)


# ------------------------------------------------------------
# Check retained observations
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("ANALYSIS-READY DATASET")
print("=" * 70)

print("Original observations :", len(airline_data))
print("Complete observations  :", len(analysis_data))
print(
    "Observations retained  :",
    round(len(analysis_data) / len(airline_data) * 100, 2),
    "%"
)

print(
    "Observations removed   :",
    len(airline_data) - len(analysis_data)
)

print("\nDate range of analysis-ready data:")
print(
    analysis_data["Date"].min(),
    "to",
    analysis_data["Date"].max()
)


# ------------------------------------------------------------
# Verify no missing values remain
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("MISSING-VALUE CHECK")
print("=" * 70)

print(
    "Total missing values in selected variables:",
    analysis_data[pca_variables].isna().sum().sum()
)

display(
    analysis_data[pca_variables].isna().sum()
)


# ------------------------------------------------------------
# Display selected variables
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("ANALYSIS DATA PREVIEW")
print("=" * 70)

display(
    analysis_data[
        ["Date", "Year", "Month"] + pca_variables
    ].head()
)
```

## STEP 8 Descriptive statistics, mean vector & covariance matrix

```python
# STEP 8: DESCRIPTIVE STATISTICS, MEAN VECTOR
#         AND COVARIANCE MATRIX


X = analysis_data[pca_variables].copy()


# ------------------------------------------------------------
# 8.1 Descriptive statistics
# ------------------------------------------------------------

print("=" * 70)
print("DESCRIPTIVE STATISTICS")
print("=" * 70)

descriptive_stats = X.describe().T

descriptive_stats = descriptive_stats[
    ["count", "mean", "std", "min", "50%", "max"]
]

descriptive_stats.columns = [
    "Count",
    "Mean",
    "Std_Dev",
    "Minimum",
    "Median",
    "Maximum"
]

display(
    descriptive_stats.round(3)
)


# ------------------------------------------------------------
# 8.2 Mean vector
# ------------------------------------------------------------

mean_vector = X.mean()

print("\n" + "=" * 70)
print("MEAN VECTOR")
print("=" * 70)

display(
    mean_vector.to_frame("Mean").round(3)
)


# ------------------------------------------------------------
# 8.3 Sample covariance matrix
# ------------------------------------------------------------

covariance_matrix = X.cov()

print("\n" + "=" * 70)
print("SAMPLE COVARIANCE MATRIX")
print("=" * 70)

display(
    covariance_matrix.round(3)
)


# ------------------------------------------------------------
# 8.4 Verify covariance matrix properties
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("COVARIANCE MATRIX CHECK")
print("=" * 70)

print(
    "Matrix dimensions:",
    covariance_matrix.shape
)

print(
    "Symmetric:",
    np.allclose(
        covariance_matrix,
        covariance_matrix.T
    )
)

print(
    "Number of variables:",
    X.shape[1]
)

print(
    "Number of observations:",
    X.shape[0]
)
```

## STEP 9 Standardization

```python
# STEP 9: STANDARDIZE VARIABLES FOR PCA


from sklearn.preprocessing import StandardScaler

# ------------------------------------------------------------
# 9.1 Extract the six PCA variables
# ------------------------------------------------------------

X = analysis_data[pca_variables].copy()


# ------------------------------------------------------------
# 9.2 Standardize using z-scores
# ------------------------------------------------------------

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

# Convert back to DataFrame for easier inspection
X_scaled_df = pd.DataFrame(
    X_scaled,
    columns=pca_variables,
    index=analysis_data.index
)


# ------------------------------------------------------------
# 9.3 Check standardized means and standard deviations
# ------------------------------------------------------------

print("=" * 70)
print("STANDARDIZATION CHECK")
print("=" * 70)

standardization_check = pd.DataFrame({
    "Mean_After_Standardization": X_scaled_df.mean(),
    "Std_After_Standardization": X_scaled_df.std(ddof=0)
})

display(
    standardization_check.round(6)
)


# ------------------------------------------------------------
# 9.4 Confirm dimensions
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STANDARDIZED DATA DIMENSIONS")
print("=" * 70)

print("Number of observations:", X_scaled.shape[0])
print("Number of variables   :", X_scaled.shape[1])


# ------------------------------------------------------------
# 9.5 Preview standardized observations
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STANDARDIZED DATA PREVIEW")
print("=" * 70)

display(
    X_scaled_df.head().round(3)
)
```

## STEP 10 Principal Component Analysis

```python
# STEP 10: PRINCIPAL COMPONENT ANALYSIS (PCA)


from sklearn.decomposition import PCA

# ------------------------------------------------------------
# 10.1 Fit PCA
# ------------------------------------------------------------

pca = PCA()

X_pca = pca.fit_transform(X_scaled)


# ------------------------------------------------------------
# 10.2 Eigenvalues
# ------------------------------------------------------------

eigenvalues = pca.explained_variance_


# ------------------------------------------------------------
# 10.3 Explained variance
# ------------------------------------------------------------

explained_variance = pca.explained_variance_ratio_ * 100

cumulative_variance = np.cumsum(
    explained_variance
)


# ------------------------------------------------------------
# 10.4 PCA summary table
# ------------------------------------------------------------

pca_summary = pd.DataFrame({
    "Principal_Component": [
        f"PC{i+1}" for i in range(len(eigenvalues))
    ],
    "Eigenvalue": eigenvalues,
    "Explained_Variance_Percent": explained_variance,
    "Cumulative_Variance_Percent": cumulative_variance
})

print("=" * 70)
print("PCA SUMMARY")
print("=" * 70)

display(
    pca_summary.round(4)
)


# ------------------------------------------------------------
# 10.5 PCA component coefficients
# ------------------------------------------------------------

component_coefficients = pd.DataFrame(
    pca.components_.T,
    index=pca_variables,
    columns=[
        f"PC{i+1}"
        for i in range(len(eigenvalues))
    ]
)

print("\n" + "=" * 70)
print("PCA COMPONENT COEFFICIENTS")
print("=" * 70)

display(
    component_coefficients.round(4)
)


# ------------------------------------------------------------
# 10.6 PCA loadings
# ------------------------------------------------------------

loadings = (
    pca.components_.T *
    np.sqrt(eigenvalues)
)

loadings_df = pd.DataFrame(
    loadings,
    index=pca_variables,
    columns=[
        f"PC{i+1}"
        for i in range(len(eigenvalues))
    ]
)

print("\n" + "=" * 70)
print("PCA LOADINGS")
print("=" * 70)

display(
    loadings_df.round(4)
)


# ------------------------------------------------------------
# 10.7 Kaiser criterion
# Retain components with eigenvalue > 1
# ------------------------------------------------------------

kaiser_components = np.sum(
    eigenvalues > 1
)

print("\n" + "=" * 70)
print("COMPONENT RETENTION")
print("=" * 70)

print(
    "Components with eigenvalue > 1:",
    kaiser_components
)

print(
    "Cumulative variance explained by first 2 components:",
    round(cumulative_variance[1], 4),
    "%"
)


# ------------------------------------------------------------
# 10.8 PCA score dimensions
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PCA SCORE MATRIX")
print("=" * 70)

print(
    "Original standardized data:",
    X_scaled.shape
)

print(
    "Full PCA score matrix:",
    X_pca.shape
)

print("\nFirst 5 PCA scores:")

display(
    pd.DataFrame(
        X_pca[:, :2],
        columns=["PC1", "PC2"],
        index=analysis_data.index
    ).head().round(4)
)
```

## STEP 11 PCA Scree Plot

```python
# STEP 11: PCA SCREE PLOT


plt.figure(figsize=(9, 5.5))

components = np.arange(1, len(eigenvalues) + 1)

plt.plot(
    components,
    eigenvalues,
    marker="o",
    linewidth=2
)

# Kaiser criterion
plt.axhline(
    y=1,
    linestyle="--",
    linewidth=1.5,
    label="Kaiser criterion (eigenvalue = 1)"
)

# Highlight retained components
plt.scatter(
    components[:2],
    eigenvalues[:2],
    s=80,
    zorder=3,
    label="Retained components"
)

plt.xticks(components)

plt.xlabel("Principal Component")
plt.ylabel("Eigenvalue")

plt.title(
    "PCA Scree Plot",
    fontsize=14,
    fontweight="bold"
)

plt.legend()
plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()
```

## STEP 12 Create the final PCA score dataset

```python
# STEP 12: CREATE FINAL PCA SCORE DATASET


# Retain the first two principal components
pca_scores = pd.DataFrame(
    X_pca[:, :2],
    columns=["PC1", "PC2"],
    index=analysis_data.index
)

# Add date and year information
pca_data = pd.concat(
    [
        analysis_data[["Date", "Year", "Month"]].reset_index(drop=True),
        pca_scores.reset_index(drop=True)
    ],
    axis=1
)

print("=" * 70)
print("FINAL PCA SCORE DATASET")
print("=" * 70)

print("Observations:", pca_data.shape[0])
print("Variables:", pca_data.shape[1])

print("\nColumns:")
print(list(pca_data.columns))

print("\nFirst 10 PCA scores:")

display(
    pca_data.head(10).round(4)
)

print("\nPCA score summary:")

display(
    pca_data[["PC1", "PC2"]]
    .describe()
    .T
    .round(4)
)
```

## STEP 13 PCA Score Plot

```python
# STEP 13: PCA SCORE PLOT


plt.figure(figsize=(10, 6))

plt.scatter(
    pca_data["PC1"],
    pca_data["PC2"],
    s=45,
    alpha=0.75
)

plt.axhline(
    y=0,
    linestyle="--",
    linewidth=1
)

plt.axvline(
    x=0,
    linestyle="--",
    linewidth=1
)

plt.xlabel(
    "PC1 — Operational Scale / Activity"
)

plt.ylabel(
    "PC2 — Utilization"
)

plt.title(
    "Observations in the PCA-Reduced Multivariate Space",
    fontsize=14,
    fontweight="bold"
)

plt.grid(
    alpha=0.25
)

plt.tight_layout()
plt.show()
```

## STEP 14 Select the number of K-means clusters

```python
# STEP 14: SELECT NUMBER OF K-MEANS CLUSTERS


from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Use the retained PCA dimensions for clustering
X_cluster = pca_data[["PC1", "PC2"]].values

k_values = range(2, 7)

silhouette_scores = []

print("=" * 70)
print("K-MEANS CLUSTER SELECTION")
print("=" * 70)

for k in k_values:

    kmeans_temp = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20
    )

    labels_temp = kmeans_temp.fit_predict(
        X_cluster
    )

    score = silhouette_score(
        X_cluster,
        labels_temp
    )

    silhouette_scores.append(score)

    print(
        f"K = {k}  |  "
        f"Silhouette Score = {score:.4f}"
    )


# ------------------------------------------------------------
# Identify best K
# ------------------------------------------------------------

best_k = list(k_values)[
    np.argmax(silhouette_scores)
]

best_score = max(silhouette_scores)

print("\n" + "=" * 70)
print("SELECTED NUMBER OF CLUSTERS")
print("=" * 70)

print("Best K:", best_k)
print("Best Silhouette Score:", round(best_score, 4))


# ------------------------------------------------------------
# Silhouette score plot
# ------------------------------------------------------------

plt.figure(figsize=(8, 5.5))

plt.plot(
    list(k_values),
    silhouette_scores,
    marker="o",
    linewidth=2
)

plt.scatter(
    best_k,
    best_score,
    s=100,
    zorder=3,
    label=f"Selected K = {best_k}"
)

plt.xticks(list(k_values))

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")

plt.title(
    "K-Means Cluster Selection Using Silhouette Score",
    fontsize=14,
    fontweight="bold"
)

plt.grid(alpha=0.25)
plt.legend()

plt.tight_layout()
plt.show()
```

## STEP 15 Fit the final K-means model

## STEP 25 Multivariate normality: Mardia's test

```python
# STEP 25: MULTIVARIATE NORMALITY — MARDIA'S TEST


from scipy.stats import chi2, norm


# ------------------------------------------------------------
# 25.1 Prepare data
# ------------------------------------------------------------

X_mvn = X_scaled.copy()

n = X_mvn.shape[0]
p = X_mvn.shape[1]


# ------------------------------------------------------------
# 25.2 Sample covariance matrix of standardized data
# ------------------------------------------------------------

S = np.cov(
    X_mvn,
    rowvar=False
)

S_inv = np.linalg.inv(S)


# ------------------------------------------------------------
# 25.3 Mahalanobis squared distances
# ------------------------------------------------------------

centered = X_mvn - X_mvn.mean(axis=0)

mahalanobis_squared = np.einsum(
    "ij,jk,ik->i",
    centered,
    S_inv,
    centered
)


# ------------------------------------------------------------
# 25.4 Mardia multivariate skewness
# ------------------------------------------------------------

# Pairwise inner products using inverse covariance
A = centered @ S_inv @ centered.T

mardia_skewness = np.mean(A ** 3)

skewness_statistic = (
    n * mardia_skewness / 6
)

skewness_df = (
    p * (p + 1) * (p + 2) / 6
)

skewness_p = chi2.sf(
    skewness_statistic,
    skewness_df
)


# ------------------------------------------------------------
# 25.5 Mardia multivariate kurtosis
# ------------------------------------------------------------

mardia_kurtosis = np.mean(
    mahalanobis_squared ** 2
)

expected_kurtosis = p * (p + 2)

# Approximate standardized z statistic
kurtosis_variance = (
    8 * p * (p + 2) / n
)

kurtosis_z = (
    mardia_kurtosis -
    expected_kurtosis
) / np.sqrt(kurtosis_variance)

kurtosis_p = (
    2 * norm.sf(abs(kurtosis_z))
)


# ------------------------------------------------------------
# 25.6 Results
# ------------------------------------------------------------

print("=" * 70)
print("MARDIA'S MULTIVARIATE NORMALITY TEST")
print("=" * 70)

print("Number of observations (n):", n)
print("Number of variables (p)   :", p)

print("\nMultivariate skewness:")
print("Statistic :", round(skewness_statistic, 4))
print("df        :", round(skewness_df, 4))
print("p-value   :", skewness_p)

print("\nMultivariate kurtosis:")
print("Statistic :", round(mardia_kurtosis, 4))
print("Expected  :", round(expected_kurtosis, 4))
print("z-value   :", round(kurtosis_z, 4))
print("p-value   :", kurtosis_p)


# ------------------------------------------------------------
# 25.7 Overall decision
# ------------------------------------------------------------

alpha = 0.05

print("\n" + "=" * 70)
print("NORMALITY DECISION")
print("=" * 70)

if skewness_p < alpha:
    print("Mardia skewness: Reject multivariate normality.")
else:
    print("Mardia skewness: Do not reject multivariate normality.")

if kurtosis_p < alpha:
    print("Mardia kurtosis: Reject multivariate normality.")
else:
    print("Mardia kurtosis: Do not reject multivariate normality.")
```

## STEP 26 Classical Mahalanobis distance

```python
# STEP 26: CLASSICAL MAHALANOBIS DISTANCE


from scipy.stats import chi2


# ------------------------------------------------------------
# 26.1 Calculate covariance matrix and its inverse
# ------------------------------------------------------------

X_mahal = X.values

mean_vector_mahal = X_mahal.mean(axis=0)

cov_matrix_mahal = np.cov(
    X_mahal,
    rowvar=False
)

cov_inverse_mahal = np.linalg.inv(
    cov_matrix_mahal
)


# ------------------------------------------------------------
# 26.2 Calculate squared Mahalanobis distance
# ------------------------------------------------------------

centered_mahal = (
    X_mahal -
    mean_vector_mahal
)

mahalanobis_d2 = np.einsum(
    "ij,jk,ik->i",
    centered_mahal,
    cov_inverse_mahal,
    centered_mahal
)


# ------------------------------------------------------------
# 26.3 Add distances to dataset
# ------------------------------------------------------------

clustered_data["Mahalanobis_D2"] = mahalanobis_d2


# ------------------------------------------------------------
# 26.4 Conventional 99% reference threshold
# ------------------------------------------------------------

alpha_mahal = 0.01

mahal_threshold = chi2.ppf(
    1 - alpha_mahal,
    df=len(pca_variables)
)

clustered_data["Mahalanobis_Reference_Flag"] = (
    clustered_data["Mahalanobis_D2"]
    > mahal_threshold
)


# ------------------------------------------------------------
# 26.5 Summary
# ------------------------------------------------------------

print("=" * 70)
print("CLASSICAL MAHALANOBIS DISTANCE")
print("=" * 70)

print("Number of observations:", len(clustered_data))
print("Number of variables:", len(pca_variables))

print(
    "\n99% chi-square reference threshold:",
    round(mahal_threshold, 4)
)

print(
    "Observations above reference threshold:",
    clustered_data[
        "Mahalanobis_Reference_Flag"
    ].sum()
)


# ------------------------------------------------------------
# 26.6 Identify highest-distance observations
# ------------------------------------------------------------

top_mahalanobis = (
    clustered_data[
        [
            "Date",
            "Year",
            "Month",
            "Cluster",
            "Regime",
            "Mahalanobis_D2",
            "Mahalanobis_Reference_Flag"
        ]
    ]
    .sort_values(
        "Mahalanobis_D2",
        ascending=False
    )
    .head(15)
    .reset_index(drop=True)
)

print("\n" + "=" * 70)
print("TOP 15 OBSERVATIONS BY MAHALANOBIS DISTANCE")
print("=" * 70)

display(
    top_mahalanobis.round(4)
)
```

## STEP 27 Robust Mahalanobis Distance using MCD

```python
# STEP 27: ROBUST MAHALANOBIS DISTANCE — MCD


from sklearn.covariance import MinCovDet


# ------------------------------------------------------------
# 27.1 Fit Minimum Covariance Determinant estimator
# ------------------------------------------------------------

mcd = MinCovDet(
    random_state=42
)

mcd.fit(X)


# ------------------------------------------------------------
# 27.2 Calculate robust squared Mahalanobis distances
# ------------------------------------------------------------

robust_mahalanobis_d2 = mcd.mahalanobis(X)


# ------------------------------------------------------------
# 27.3 Add robust distances to dataset
# ------------------------------------------------------------

clustered_data["Robust_Mahalanobis_D2"] = (
    robust_mahalanobis_d2
)


# ------------------------------------------------------------
# 27.4 Empirical 99th-percentile cutoff
# ------------------------------------------------------------

robust_threshold = np.percentile(
    robust_mahalanobis_d2,
    99
)

clustered_data["Robust_Anomaly_Flag"] = (
    clustered_data["Robust_Mahalanobis_D2"]
    > robust_threshold
)


# ------------------------------------------------------------
# 27.5 Summary
# ------------------------------------------------------------

print("=" * 70)
print("ROBUST MAHALANOBIS DISTANCE — MCD")
print("=" * 70)

print("Number of observations:", len(clustered_data))
print("Number of variables:", len(pca_variables))

print(
    "\nEmpirical 99th-percentile threshold:",
    round(robust_threshold, 4)
)

print(
    "Observations above robust threshold:",
    clustered_data[
        "Robust_Anomaly_Flag"
    ].sum()
)


# ------------------------------------------------------------
# 27.6 Identify highest robust distances
# ------------------------------------------------------------

top_robust = (
    clustered_data[
        [
            "Date",
            "Year",
            "Month",
            "Cluster",
            "Regime",
            "Mahalanobis_D2",
            "Robust_Mahalanobis_D2",
            "Robust_Anomaly_Flag"
        ]
    ]
    .sort_values(
        "Robust_Mahalanobis_D2",
        ascending=False
    )
    .head(15)
    .reset_index(drop=True)
)

print("\n" + "=" * 70)
print("TOP 15 OBSERVATIONS BY ROBUST MAHALANOBIS DISTANCE")
print("=" * 70)

display(
    top_robust.round(4)
)
```

## STEP 28 Compare classical and robust anomaly detection

```python
# STEP 28: CLASSICAL VS ROBUST ANOMALY COMPARISON


# ------------------------------------------------------------
# 28.1 Create comparison table
# ------------------------------------------------------------

anomaly_comparison = clustered_data[
    [
        "Date",
        "Year",
        "Month",
        "Cluster",
        "Regime",
        "Mahalanobis_D2",
        "Mahalanobis_Reference_Flag",
        "Robust_Mahalanobis_D2",
        "Robust_Anomaly_Flag"
    ]
].copy()


# ------------------------------------------------------------
# 28.2 Keep observations flagged by either method
# ------------------------------------------------------------

flagged_comparison = anomaly_comparison[
    (
        anomaly_comparison["Mahalanobis_Reference_Flag"]
        |
        anomaly_comparison["Robust_Anomaly_Flag"]
    )
].sort_values(
    "Date"
).reset_index(drop=True)


print("=" * 70)
print("CLASSICAL VS ROBUST MAHALANOBIS COMPARISON")
print("=" * 70)

display(
    flagged_comparison.round(4)
)


# ------------------------------------------------------------
# 28.3 Summary counts
# ------------------------------------------------------------

classical_count = (
    clustered_data["Mahalanobis_Reference_Flag"]
    .sum()
)

robust_count = (
    clustered_data["Robust_Anomaly_Flag"]
    .sum()
)

both_count = (
    clustered_data[
        clustered_data["Mahalanobis_Reference_Flag"]
        &
        clustered_data["Robust_Anomaly_Flag"]
    ]
    .shape[0]
)

classical_only = (
    clustered_data[
        clustered_data["Mahalanobis_Reference_Flag"]
        &
        ~clustered_data["Robust_Anomaly_Flag"]
    ]
    .shape[0]
)

robust_only = (
    clustered_data[
        ~clustered_data["Mahalanobis_Reference_Flag"]
        &
        clustered_data["Robust_Anomaly_Flag"]
    ]
    .shape[0]
)


print("\n" + "=" * 70)
print("ANOMALY DETECTION SUMMARY")
print("=" * 70)

print(
    "Classical Mahalanobis flags:",
    classical_count
)

print(
    "Robust MCD flags:",
    robust_count
)

print(
    "Flagged by both methods:",
    both_count
)

print(
    "Classical-only flags:",
    classical_only
)

print(
    "Robust-only flags:",
    robust_only
)


# ------------------------------------------------------------
# 28.4 Final robust extreme observations
# ------------------------------------------------------------

final_robust_extremes = clustered_data[
    clustered_data["Robust_Anomaly_Flag"]
][
    [
        "Date",
        "Year",
        "Month",
        "Cluster",
        "Regime",
        "Mahalanobis_D2",
        "Robust_Mahalanobis_D2"
    ]
].sort_values(
    "Robust_Mahalanobis_D2",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 70)
print("FINAL ROBUST MULTIVARIATE EXTREMES")
print("=" * 70)

display(
    final_robust_extremes.round(4)
)
```

## STEP 29 Temporal trend of the PCA dimensions and relative cargo activity

```python
# ============================================================
# STEP 29: TEMPORAL TREND ANALYSIS — CORRECTED
# ============================================================

# ------------------------------------------------------------
# 29.1 Add PCA scores to clustered_data
# ------------------------------------------------------------

clustered_data["PC1"] = pca_data["PC1"].values
clustered_data["PC2"] = pca_data["PC2"].values


# ------------------------------------------------------------
# 29.2 Convert dates to a numeric time index
# ------------------------------------------------------------

time_index = (
    clustered_data["Date"]
    .map(pd.Timestamp.toordinal)
)


# ------------------------------------------------------------
# 29.3 Variables to test
# ------------------------------------------------------------

trend_variables = {
    "PC1": clustered_data["PC1"],
    "PC2": clustered_data["PC2"],
    "Relative_Cargo_Activity":
        clustered_data["Relative_Cargo_Activity"]
}


# ------------------------------------------------------------
# 29.4 Spearman trend tests
# ------------------------------------------------------------

trend_results = []

for variable_name, values in trend_variables.items():

    rho, p_value = spearmanr(
        time_index,
        values
    )

    trend_results.append({
        "Variable": variable_name,
        "Spearman_Rho": rho,
        "P_Value": p_value
    })


trend_results_df = pd.DataFrame(
    trend_results
)


# ------------------------------------------------------------
# 29.5 Display results
# ------------------------------------------------------------

print("=" * 70)
print("TEMPORAL TREND ANALYSIS")
print("=" * 70)

display(
    trend_results_df.round(4)
)


# ------------------------------------------------------------
# 29.6 Significance interpretation
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TREND SIGNIFICANCE")
print("=" * 70)

for _, row in trend_results_df.iterrows():

    if row["P_Value"] < 0.05:
        significance = "Statistically significant"
    else:
        significance = "Not statistically significant"

    print(
        f'{row["Variable"]}: '
        f'rho = {row["Spearman_Rho"]:.4f}, '
        f'p = {row["P_Value"]:.4f} '
        f'→ {significance}'
    )
```
